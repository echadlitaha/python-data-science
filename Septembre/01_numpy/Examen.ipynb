Partie B — Pandas

```python
commandes = pd.DataFrame({
    "Commande_ID": ["C001", "C002", "C003", "C004", "C004", "C005", "C006"],
    "Client_ID": [1, 2, 1, 3, 3, 4, 5],
    "Ville": [" casa ", "RABAT", "Casa", "Marrakech", "Marrakech", "rabat", "Casa"],
    "Produit": ["PC", "Telephone", "Ecran", "PC", "PC", "Telephone", "Ecran"],
    "Montant": [8000, 4500, np.nan, 7000, 7000, 5200, 3000],
    "Quantite": [1, 2, 2, 1, 1, 3, 2],
    "Date_Commande": [
        "01/09/2026",
        "03/09/2026",
        "05/09/2026",
        "07/09/2026",
        "07/09/2026",
        "10/09/2026",
        "12/09/2026"
    ]
})

clients = pd.DataFrame({
    "Client_ID": [1, 2, 3, 4, 6],
    "Nom": ["Ali", "Sara", "Omar", "Lina", "Nadia"]
})

#7
commandes.isna().sum()
commandes.duplicated()
#8
commandes["Ville"] = commandes["Ville"].str.strip().str.title()
commandes.loc[:,["Ville"]]
commandes = commandes.drop_duplicates()
#9
commandes["Date_Commande"] = pd.to_datetime(
    commandes["Date_Commande"],
    format="%d/%m/%Y"
)
commandes["Mois"] = commandes["Date_Commande"].dt.month
commandes["Jour"] = commandes["Date_Commande"].dt.day
#10
commandes["Montant"] = commandes["Montant"].fillna(
    commandes["Montant"].median()
)
#11
commandes.loc[
    commandes["Ville"].isin(["Casa", "Rabat"])
    & (commandes["Montant"] >= 5000),
    ["Commande_ID", "Ville", "Produit", "Montant"]
]
#12
commandes["CA"] = commandes["Montant"] * commandes["Quantite"]
rapport = commandes.groupby("Ville").agg(
    CA_Total=("CA", "sum"),
    Montant_Moyen=("Montant", "mean")
).reset_index()
#13
pd.merge(commandes,clients, on = "Client_ID" , how = "left")
# c'est client numero 5 car il n'exite pas dans la DataFrame clients
#14
pd.pivot_table(
    commandes,
    values="CA",
    index="Ville",
    columns="Produit",
    aggfunc="sum",
    fill_value=0
)
#15
# merge et le suffixes
#16
""" remplacer une valeur manquante par une statistique comme la moyenne ou la médiane plus
utile que supprimer les lignes manquantes car la moyenne et surtout la mediane donnent un chiffre
approxiamtif a le chiffre manquant """
#17
""" je choisi la mediane car la valeur 50 000 est tres loin de 4000 et 8000"""
#18
""" commandes["Montant"] >= 5000 : donne la plage booleen 
commandes[commandes["Montant"] >= 5000] : donne seulement les lignes ou le montant est supetrieures a
 5000 """
```
