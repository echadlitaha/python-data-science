```python
#Jour5 
# Janvier | fevrier | mars
ventes = np.array([
    [1200, 1500, 1700],
    [2000, 2100, 2300],
    [900,  np.nan, 1200],
    [1800, 1900, 2000],
    [2500, 2700, np.nan],
    [1400, 1600, 1800]
])
categories = np.array([
    "A", "B", "A", "C", "B", "A"
])
#Partie 1 — Valeurs manquantes
#1
print(np.isnan(ventes))
#2
print(np.isnan(ventes).sum())
#3
print(np.nanmean(ventes))
#4
print(np.nansum(ventes))
```

```python
#Partie 2 :
#5
print(np.nansum(ventes,axis = 0))
#6
moyenne = (np.nanmean(ventes,axis=1)).round(2)
print(moyenne)
#Partie 3:
#7
print(np.where(moyenne>=1800,'fort','normal'))
#8
print(np.unique(categories))
#9
categorie,compter = np.unique(categories,return_counts=True)
print(categorie)
print(compter)
```

```python
#Partie 4 : Assemblage
nouvelles_ventes = np.array([
    [1600, 1700, 1900],
    [2200, 2400, 2600]
])
#10
ventes_completes = np.concatenate((ventes,nouvelles_ventes))
print(ventes_completes)
#11
print(ventes_completes.shape)

#Partie 5 — split()
x = np.array([10, 20, 30, 40, 50, 60, 70, 80])
#12
print(np.split(x,4))

#Partie 6 — np.random
#13
np.random.seed(42)
matrice = np.random.randint(100,500,size=(3, 4))
print(matrice)
```
