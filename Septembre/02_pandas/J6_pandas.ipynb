```python
#Jour 7 : Selection et filtres 
data = {
    "Nom": ["Ali", "Sara", "Omar", "Lina", "Youssef"],
    "Age": [25, 30, 28, 35, 40],
    "Ville": ["Casa", "Rabat", "Casa", "Marrakech", "Rabat"],
    "Salaire": [4000, 5500, 4800, 6500, 7200]
}

df = pd.DataFrame(data)
df
df.index = ["E101", "E102", "E103", "E104", "E105"]
#1
df.loc["E103","Salaire"]
#2
df.loc["E105"]
#3
df.loc["E102":"E104"]
#4
df.loc["E102":"E104",["Nom","Salaire"]]
#5
df.iloc[2]
#6
df.iloc[4,1]
#7
#parceque .loc utilse labelle pour les lignes , puisque on a changer l'index donc 2 n'existe plus
#8
#parceque .iloc ne regarde pas sur les noms , elle donne les positions numrqiues des lignes comme des colonnes 
```

```python
data = {
    "Client": ["Ali", "Sara", "Omar", "Lina", "Youssef", "Salma", "Amine", "Nadia"],
    "Ville": ["Casa", "Rabat", "Casa", "Marrakech", "Rabat", "Casa", "Tanger", "Marrakech"],
    "Produit": ["PC", "Telephone", "Tablette", "PC", "Ecran", "Telephone", "PC", "Tablette"],
    "Categorie": ["Informatique", "Mobile", "Mobile", "Informatique",
                  "Informatique", "Mobile", "Informatique", "Mobile"],
    "Montant": [8000, 4500, 3200, 9000, 2500, 5000, 7500, 3800],
    "Quantite": [1, 2, 1, 1, 2, 3, 1, 2]
}

df = pd.DataFrame(data)
df
#1
df.loc[:,["Client","Produit","Montant"]]
#2
df.loc[2:5,]
#3
df.iloc[0:4,0:3]
#4
df[df["Montant"] > 5000]
#5
df[df["Ville"].isin(["Casa","Rabat"])]
#6
df[(df["Categorie"] == "Informatique") & (df["Montant"] >= 7000)]
#7
df[df["Montant"].between(3500,6000)]
#8
df.loc[df["Quantite"] >= 2 ,["Client","Ville","Montant"]]
#9
df[df["Produit"].isin(["PC","Telephone","Ecran"])]
#10
df[(df["Ville"] == "Casa") & (df["Quantite"] > 1)]
#11
df[(df["Ville"] == "Marrakech") | (df["Montant"] > 7000)]
#12
df.iloc[5,3]
#13
df[(df["Ville"]=="Rabat") & (df["Montant"] > 3000)]
#14
df.loc[(df["Ville"].isin(["Casa","Marrakech"])) & (df["Montant"].between(3000,8000)),["Client","Produit","Montant"]]
#15
#.loc prendre les lignes et les colonnes dous forme de labels et noms par contre .iloc le prendre sous forme de position numerique
```
