```python
#Exercice:
import pandas as pd

data = {
    "Vendeur": ["Ali", "Sara", "Ali", "Omar", "Sara", "Ali", "Omar", "Sara"],
    "Ville": ["Casa", "Rabat", "Casa", "Casa", "Rabat", "Rabat", "Casa", "Casa"],
    "Produit": ["PC", "Telephone", "Ecran", "PC", "PC", "Telephone", "Ecran", "PC"],
    "Montant": [8000, 4500, 2000, 7500, 8200, 5000, 2200, 7800],
    "Quantite": [1, 2, 2, 1, 1, 3, 2, 1]
}
df = pd.DataFrame(data)
#1
pd.crosstab(df["Ville"],df["Produit"],values=df["Montant"],aggfunc="sum",margins = True)
#2
pd.crosstab(df["Ville"],df["Produit"],values=df["Montant"],aggfunc="sum",margins = True).fillna(0)
#3
pd.crosstab(df["Ville"],df["Produit"],values=df["Montant"],aggfunc="mean",margins = True).fillna(0)
#4
pd.pivot_table(df,values = "Quantite",index = "Vendeur",columns = "Produit",aggfunc = "sum")
#5
pd.pivot_table(df,values = "Montant",index = ["Ville","Vendeur"],columns = "Produit",aggfunc = "sum")
#6
pd.pivot_table(df,values = "Montant",index = "Ville",columns = "Produit",aggfunc = ["sum","mean"])
#7
pd.pivot_table(df,values = "Montant",index = "Ville",columns = "Produit",aggfunc = "sum",fill_value = 0,margins=True)
#8
pd.crosstab(df["Ville"],df["Produit"])
#9
pd.crosstab(df["Ville"],df["Produit"],margins=True)
#10
"""groupby : Somme et moyenne par groupes
pivot_table : Tableau de synthèse lignes * colonnes"""
#11
""" pivot_table : Tableau de synthèse lignes * colonnes
crosstab : Compter les combinaisons de catégories"""
#12
"""values : la valeur qu'on va la calculer 
 index: les lignes
columns: les colonnes 
aggfunc : operation mathematique : sum ou mean ou min ou max ..."""
```
