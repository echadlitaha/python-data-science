J12-	GroupBy & agrégations

```python
#Exercice 1:
import pandas as pd

data = {
    "Vendeur": ["Ali", "Sara", "Ali", "Omar", "Sara", "Ali", "Omar", "Sara"],
    "Ville": ["Casa", "Rabat", "Casa", "Casa", "Rabat", "Rabat", "Casa", "Casa"],
    "Produit": ["PC", "Telephone", "Ecran", "PC", "PC", "Telephone", "Ecran", "PC"],
    "Montant": [8000, 4500, 2000, 7500, 8200, 5000, 2200, 7800],
    "Quantite": [1, 2, 2, 1, 1, 3, 2, 1]
}

df = pd.DataFrame(data)
#1
df.groupby("Vendeur")["Montant"].sum()
#2
df.groupby("Vendeur")["Montant"].mean().round(2)
#3
df.groupby("Vendeur")["Quantite"].sum()
#4
df.groupby("Vendeur")["Montant"].max()
#5
df.groupby("Vendeur")["Montant"].min()
#6
df.groupby("Vendeur")["Montant"].count()
#7
df.groupby("Vendeur")["Montant"].size()
#8
df.groupby("Ville")["Montant"].sum()
#9
df.groupby("Ville")["Montant"].mean()
#10
df.groupby("Produit")["Quantite"].sum()
#11
df.groupby("Vendeur")["Montant"].sum().reset_index()
#12
""" df["Montant"].sum() retourne la somme du montant de tous les vendeur en une seule ligne  
df.groupby("Vendeur")["Montant"].sum() retourne la somme du montant de chaque vendeur en des groupes """
```

```python
#Exercice 2:
data = {
    "Vendeur": ["Ali", "Sara", "Ali", "Omar", "Sara", "Ali", "Omar", "Sara"],
    "Ville": ["Casa", "Rabat", "Casa", "Casa", "Rabat", "Rabat", "Casa", "Casa"],
    "Produit": ["PC", "Telephone", "Ecran", "PC", "PC", "Telephone", "Ecran", "PC"],
    "Montant": [8000, 4500, 2000, 7500, 8200, 5000, 2200, 7800],
    "Quantite": [1, 2, 2, 1, 1, 3, 2, 1]
}

df = pd.DataFrame(data)
#1
df.groupby(["Vendeur","Ville"]).agg(montant_total = ("Montant","sum"))
#2
df.groupby(["Vendeur","Ville"]).agg(montant_total = ("Montant","sum")).reset_index()
#3
df.groupby("Vendeur")["Montant"].agg(["sum","mean","min","max"]).reset_index()
#4
df.groupby("Vendeur").agg({"Montant" : ["sum"],"Quantite":["sum"]}).reset_index()
#5
df.groupby("Ville").agg({"Montant" :["sum","mean"],"Quantite" : ["sum"]}).reset_index()
#6
df.groupby(["Ville", "Produit"]).agg(
    Montant_Total=("Montant", "sum")
).reset_index()
#7
df.groupby("Vendeur").agg(
 CA_Total = ("Montant","sum"),
 Vente_Moyenne =("Montant","mean"),
 Quantite_Total = ("Quantite","sum")).reset_index()
#8
df.groupby("Vendeur").agg(
 CA_Total = ("Montant","sum"),
 Vente_Moyenne =("Montant","mean"),
 Quantite_Totale = ("Quantite","sum"),
 Nb_Ventes = ("Montant","count")).reset_index()
#9
df.groupby("Ville").agg(
 CA_Total = ("Montant","sum"),
 Vente_Min =("Montant","min"),
 Vente_Max = ("Montant","max"),
 Quantite_Totale = ("Quantite","sum")).reset_index()
#10
''' groupby("Vendeur") grouper par une seule colonne Vendeur
groupby(["Vendeur", "Ville"]) grouper par deux colonnes Vendeur et Ville '''
#11
#.agg() permet d'appliquer plusieurs fonctions d'agrégation en une seule opération, éventuellement sur plusieurs colonnes.
#12
''' CA_Total est un alias de nouveau colonne
Montant est une colonne
sum operation '''
```
