E- Dates

```python
#27
ventes["Date_Commande"] = pd.to_datetime(ventes["Date_Commande"],
    format = "%d/%m/%Y")
#28
ventes["Annee"] = ventes["Date_Commande"].dt.year
ventes["Mois"] = ventes["Date_Commande"].dt.month
ventes["Jour"] = ventes["Date_Commande"].dt.day
#29
ventes[ventes["Date_Commande"] >= "2026-09-10"]
#30
ventes["Date_Commande"].min()
ventes["Date_Commande"].max()
#31
ventes["Jours_Depuis_Premiere"] = (
    ventes["Date_Commande"]
    - ventes["Date_Commande"].iloc[0]
).dt.days
```
