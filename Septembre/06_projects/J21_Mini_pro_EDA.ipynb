Partie 2 : Nettoyage + analyse Pandas/NumPy

```python
#Etape A:
#1
df["Ville"] = df["Ville"].str.strip().str.title()
#2
df["Categorie"] = df["Categorie"].str.strip().str.title()
#3
df["Date"]=pd.to_datetime(df["Date"], format = "%d/%m/%Y")
#4
"""Les valeurs des Remise_Pct et de Satisfaction sont tres rapproches entre eux, 
donc pour les valeurs manquantes on doit les remplacer avec le mean """

df["Remise_Pct"] = pd.to_numeric(df["Remise_Pct"], errors='coerce')
df["Remise_Pct"] = df["Remise_Pct"].fillna(df["Remise_Pct"].median())

df["Satisfaction"] = pd.to_numeric(df["Satisfaction"], errors='coerce')
df["Satisfaction"] = df["Satisfaction"].fillna(df["Satisfaction"].median())
#5
df[df.duplicated()]
#6
"""On constate que la valeur Quantite = 20 est tres atyptiques, ici il y a la presence du mediane !
mais on ne doit pas la supprimer, car c'est une vrai donnee viens de notre Dataset , donc supprimer
Quantite = 20 il nous donne un probleme"""
#7
df.info()
df.isna().sum()
df.duplicated().sum()
df.nunique()
#Etape B :
df["CA_Brut"] = df["Quantite"] * df["Prix_Unitaire"]

df["CA_Net"] = df["CA_Brut"] * (1 - (df["Remise_Pct"] / 100))
#Etape C :
df_resultat = (
    df.groupby("Categorie")["Quantite"].sum().sort_values(ascending=False).reset_index())
df_resultat.head(1)

df_resultat2 = (df.groupby("Categorie")["CA_Net"].sum().sort_values(ascending=False).reset_index())
df_resultat2.head(1)
```
