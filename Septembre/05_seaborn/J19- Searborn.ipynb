Exercice : Heure 2

```python
df = pd.DataFrame({
    "Service": [
        "IT", "IT", "IT", "IT", "IT",
        "Finance", "Finance", "Finance", "Finance",
        "RH", "RH", "RH", "RH", "RH"
    ],
    "Salaire": [
        6500, 7000, 7200, 7600, 14000,
        6000, 6400, 6800, 7200,
        4300, 4500, 4700, 4900, 5200
    ],
    "Experience": [
        2, 3, 5, 6, 8,
        2, 4, 6, 8,
        1, 2, 3, 4, 6
    ]
})

#1
sns.histplot(data=df,x="Salaire")
#2
sns.histplot(data=df,x="Salaire",bins=6)
#3
sns.histplot(data=df, x="Salaire", bins=6, kde = True)
#4
"""On remarque que les salaires entre 4300 et 7600 Dh sont approches, avec une valeur de 14000 dh
qui est tres elevees qui crée une asymétrie vers la droite."""
#5 
sns.boxplot(data=df, y = "Salaire")
#6
sns.boxplot(data=df, x="Service", y="Salaire")
#7
"""D'apres le boxplot le service qui contient une valeur atypique élevée est : IT"""
#8
sns.countplot(data=df, x="Service")
#9
"""Ce graphique montre e nombre d'employés dans chaque service.
Le barplot montre la moyenne des salaires par service."""
#10
"""A-distribution des âges : histplot
B-nombre de clients par ville : countplot
C-salaire moyen par département : barplot
D-comparaison de la dispersion des salaires entre départements : boxplot
E-détection visuelle de valeurs atypiques : boxplot"""
```

Exercice de synthèse final

```python
df = pd.DataFrame({
    "Service": [
        "IT", "IT", "IT", "IT",
        "Finance", "Finance", "Finance", "Finance",
        "RH", "RH", "RH", "RH"
    ],

    "Experience": [
        2, 4, 6, 9,
        1, 3, 5, 8,
        1, 2, 4, 6
    ],

    "Salaire": [
        6000, 7200, 8500, 10500,
        5000, 6000, 7000, 9000,
        4200, 4700, 5400, 6200
    ],

    "Performance": [
        60, 70, 80, 92,
        55, 65, 75, 88,
        50, 58, 68, 78
    ],

    "Age": [
        26, 30, 34, 39,
        24, 28, 32, 37,
        23, 27, 31, 35
    ]
})
#1
sns.scatterplot(data=df, x="Experience", y="Salaire")
#2
sns.scatterplot(data=df, x="Experience", y="Salaire", hue="Service")
#3
sns.barplot(data=df,x="Service",y="Salaire")
#4
sns.countplot(data=df, x="Service")
#5
sns.histplot(data=df, x="Salaire")
#6
sns.histplot(data=df, x="Salaire", kde=True )
#7
sns.boxplot(data=df,x="Service",y="Salaire")
#8
"""Pour la question 3 le barplot il ne montre pas la dispersion, ni les outliers
par contre le boxplot montre la forme de distribution et les outliers"""
#9
corr = df.corr(numeric_only=True)
#10
sns.heatmap(corr, annot=True, fmt = ".2f")
#11
"""D'apres le heatmap() le relation semble la plus forte avec Salaire est :
La performance, 0.93"""
#12
"""Car il y a la symetrie : Cor(A,B) = Cor(B,A) donc on doit prendre une seul entre ces deux"""
#13
"""Car en math Cor(A,A) = 1 toujour"""
#14
"""Premierement , on ne doit pas dire cause directement , car on ne sait pas s'il y a une 
autre valeur ou donnee qui influence ces deux ,
deuxiement , puisque corr > 0.8 donc il y a une forte relation lineaire positive"""
#15
"""A- évolution d'une variable: lineplot()
B- relation entre deux variables numériques : scatterplot()
C- comparaison d'une moyenne par catégorie : barplot()
D- nombre d'observations par catégorie: countplot()
E- distribution d'une variable numérique : histplot()
F- comparaison des distributions et valeurs atypiques : boxplot()
G- vue générale des corrélations entre variables numériques: heatmap()"""
```

```python
#1 
ca_mensuel = (
    df.groupby("Mois", sort=False, as_index=False)["CA"].sum())
sns.lineplot(data=ca_mensuel,x="Mois",y="CA",marker="o")
#2
# Le mois est juin
#3
sns.barplot(data = df , x="Categorie",y="CA")
#4
#la catégorie ayant le chiffre d'affaires moyen le plus élevé est le PC
#5 
sns.boxplot(data = df, x= "Ville",y="CA")
#6
"""Ce graphe permet de comparer visuellement la mediane , la dispersion .
On constate que mediane du Casablanca = celle du Rabat , ainsi on constate que Rabat a une valeur
outlier ( CA = 46000)"""
#7
sns.scatterplot(data=df, x="Budget_Publicite",y="CA", hue="Categorie")
#8
"""On constate d'apres la nuage du points qu'il ya une forte tendance postive du bas a gauche 
jusqu'au haut a droite , plus le CA augmente plus le budget_publicitaire augmente aussi"""
#9
sns.histplot(data=df, x="Benefice")
#10
"""On constate que tous les valeurs sont rapprochees les unes des autres sauf les valeurs entre 
~ 4500 et ~5500 qui sont tres hautes , et les valeurs entre ~7500 et ~8500 qui sont tres basses"""
#11
sns.boxplot(data=df, x="Categorie" ,y="Benefice")
#12
corr = df.corr(numeric_only= True)
sns.heatmap(corr,annot=True)
#13
"""Budget_Publicite et Benefice , les deux ces correlations avec CA = 1"""
#14
"""Je ne suis pas d'accord avec le responsable , car meme si il y a une forte correlation , on ne sait pas
s'il y a une autre valeur ou donnee qui influence ces deux (Le budget publicitaire et le chiffre d'affaires)"""
```

```python
#15
fig,axes = plt.subplots(2,2,figsize=(10,8))

sns.lineplot(data = df , x="Mois",y="CA",marker = "o", ax=axes[0,0])
axes[0,0].set_title("le lineplot")
axes[0,0].grid()

sns.barplot(data = df , x="Categorie",y="CA",ax=axes[0,1])
axes[0,1].set_title("le barplot")

sns.boxplot(data = df, x= "Ville",y="CA",ax=axes[1,0])
axes[1,0].set_title("le boxplot")

sns.scatterplot(data=df, x="Budget_Publicite",y="CA", hue="Categorie",ax=axes[1,1])
axes[1,1].set_title("le scatterplot")
axes[1,1].grid()

fig.suptitle("Les 4 graphiques du quest 15")
plt.tight_layout(rect=[0,0,1,0.95])
plt.show()

#16
"""Le chiffre d'affaires total augmente globalement sur les six mois, passant de 100 000 DH en janvier à 150 500 DH 
en juin.
La catégorie PC possède le CA moyen le plus élevé, suivie du téléphone.
Casablanca et Rabat ont la même médiane de CA, tandis que Rabat présente une observation particulièrement 
élevée à 46 000 DH.
Le budget publicitaire présente une très forte association positive avec le chiffre d'affaires.
Le bénéfice est la variable numérique la plus fortement corrélée au CA, 
mais ces corrélations ne permettent pas à elles seules d’établir une causalité."""
```
