```python
prix = np.array([
    100, 105, 110, 110, 115,
    120, 125, 130, 135, 300
])
#1
variance = np.var(prix,ddof=0)
#2
ecart_type = np.std(prix,ddof=0)
ecart_type.round(2)
#3
Q1 = np.quantile(prix,0.25)
Q3 = np.quantile(prix,0.75)
#4
IQR = Q3 - Q1
#5
borne_basse = Q1 - (1.5 * IQR)
borne_haute = Q3 + (1.5 * IQR)
#6
valeurs_atypiques = prix[(prix < borne_basse) | (prix > borne_haute)]
#7
""" car la valeur 300 est tres atyptiques des autres valeurs , les 9 premieres valeurs sont
tres proches de l'un l'autre , mais 300 est tres loin"""
#8
""" va diminuer car les valeurs sont tres proches maintenant apres la suppression de 300"""
#9
""" Je dois d’abord vérifier si cette valeur est correcte avant de décider de la conserver ou de la supprimer."""
#10
""" IQR est le moins sensibles a les valeurs extremes car il est baser sur les quantiles , 
mais l'etendue est le plus sensibles car il se base sur le max et le min"""
```

J16 — Corrélation et relations entre variables

Exercice : Heure 1

```python
heures = np.array([1, 2, 3, 4, 5, 6, 7, 8])
notes = np.array([8, 9, 10, 12, 13, 15, 16, 18])
prix = np.array([20, 25, 30, 35, 40, 45, 50, 55])
quantites = np.array([120, 110, 105, 90, 80, 75, 60, 50])

#1
cov = np.cov(heures,notes)
cov[0,1]
#2
""" Covariance = 8.64 > 0 donc c'est positive 
heures et notes s'evoluent en meme sens """
#3
correlation = np.corrcoef(heures,notes)
correlation[0,1]
#4
""" il y a une relation lineaire tres forte positivement"""
#5
cova = np.cov(prix,quantites)
cova[0,1]
#6
corr = np.corrcoef(prix,quantites)
corr[0,1]
#7
""" On remarque que prix et quantites s'evoluent dans le sens opposer
ainsi il y a une relation lineaire tres forte negativement"""
#8
""" Non , car on ne sait pas s'il y a une troisieme valeur qui a influncer ce calcule
par exemple il peut etre une colonnes de situation mentale ou familiale"""
#9
""" le - signifie la direction du correlation 
le 0.87 signifie la force , ici c'est tres forte (>0.8)"""
#10
""" Il n y'a pas ou peu de relation lineaire """
```
