Exercice de synthèse final

```python
import pandas as pd

df = pd.DataFrame({
    "Mois": ["Jan", "Fev", "Mar", "Avr", "Mai", "Juin"],
    "CA": [12000, 14500, 13800, 17000, 19500, 23000],
    "Benefice": [2500, 3200, 2800, 4100, 5000, 6200]
})

produits = pd.DataFrame({
    "Produit": ["PC", "Telephone", "Ecran", "Tablette"],
    "Quantite": [85, 140, 65, 100]
})

employes = pd.DataFrame({
    "Experience": [1, 2, 3, 5, 7, 9, 12, 15],
    "Salaire": [4200, 4600, 5000, 5900, 6800, 7800, 9500, 11000]
})

#1
plt.plot(df["Mois"],df["CA"])
#2
plt.plot(df["Mois"],df["CA"],marker="o")
plt.grid()
#3
plt.plot(df["Mois"],df["CA"],marker="o",label="CA")
plt.title("Evolution mensuelle du CA")
plt.xlabel("Mois")
plt.ylabel("CA")
plt.legend()
#4
plt.plot(df["Mois"],df["CA"],marker="o",label="CA")
plt.plot(df["Mois"],df["Benefice"],marker="o",label="Benefice")
plt.legend()
#5
"""Le mois est mars a diminuer par rapport au fevrier"""
#6
plt.bar(produits["Produit"],produits["Quantite"])
#7
plt.title("Quantités vendues par produit")
plt.xlabel("Les 4 Produits")
plt.ylabel("Les quantites")
#8
# le produit est telephone : 140
#9
plt.scatter(employes["Experience"],employes["Salaire"])
#10
plt.title("La relation entre l'experience et le salaire d'un employer")
plt.xlabel("Experience")
plt.ylabel("Salaire")
plt.grid()
#11
"""D'apres le graphe scatter je constate qu'il y a une forte tendance positive entre experience et 
salaire"""
#12
"""Pour comparer les categories , le bar c'est l'outil principale , et étudier la relation expérience/salaire
qui sont deux valeurs numeriques on doit utiliser le scatter"""
#13
"""Une courbe qui reliant tous les points est utiliser quand il y a une evolution / Tendance,
donc le choix du collegue est moins approprié """
#14
"""les problèmes que cela va poser est quand par exemple un person qui ne sait pas le code il n'aura
comprend rien , et meme si le collegue lui meme quand il voudra extraire une donner dans le graphe
il ne pourra comprend rien """
#15
"""1- plt.plot : evolution / tendence , temps(quotidienne en mois) 
2- plt.bar : comparaison par categorie
3- plt.scatter : relation entre deux valeurs numeriques
4- plt.plot : evolution annuelle 
5- plt.bar : comparaison par categorie"""
```

3. Deux graphiques côte à côte

```python
mois = ["Jan", "Fev", "Mar", "Avr", "Mai"]
ca = [12000, 14500, 13800, 17000, 19500]
benefice = [2500, 3200, 2800, 4100, 5000]

fig, axes = plt.subplots(1, 2)

axes[0].plot(mois, ca)
axes[0].set_title("Chiffre d'affaires")
axes[0].set_xlabel("Mois")
axes[0].set_ylabel("CA")

axes[1].plot(mois, benefice)
axes[1].set_title("Bénéfice")
axes[1].set_xlabel("Mois")
axes[1].set_ylabel("Bénéfice")

plt.show()
```

Exercice : Heure 1

```python
mois = ["Jan", "Fev", "Mar", "Avr", "Mai", "Juin"]
ca = [12000, 14500, 13800, 17000, 19500, 23000]
benefice = [2500, 3200, 2800, 4100, 5000, 6200]

produits = ["PC", "Telephone", "Ecran", "Tablette"]
quantites = [85, 140, 65, 100]

experience = [1, 2, 3, 5, 7, 9]
salaire = [4200, 4600, 5000, 5900, 6800, 7800]

#1
fig,axes=plt.subplots(1,2)
#2
axes[0].plot(mois,ca)
#3
axes[1].plot(mois,benefice)
#4
axes[0].set_title("evolution de CA")
axes[1].set_title("evolution du benefice")
#5
fig.suptitle("evolution du CA et benefice")
#6
fig,axes = plt.subplots(2,1)
axes[0].bar(produits,quantites)
axes[1].scatter(experience,salaire)
#7
axes[0].set_title("Comparaison quantites par produit")
axes[1].set_title("Relation entre experience et salaire")

axes[0].set_xlabel("Produit")
axes[0].set_ylabel("Quantite")

axes[1].set_xlabel("Experience")
axes[1].set_ylabel("Salaire")
plt.tight_layout()
plt.show()
#8
fig,axes = plt.subplots(2,2)
""" en haut à gauche : axes[0,0]
en haut à droite : axes[0,1]
en bas à gauche : axes[1,0]
en bas à droite : axes[1,1]"""
#9
"""fig
→ toute la figure / la feuille complète
axes
→ un ou plusieurs graphiques placés dans cette figure """
#10
"""Dans le cas lorsqu’on utilise plusieurs subplots , il peut avoir un chevauchement ou 
des titre coller sans espace avec les autres , donc on utilise tight_layout() pour
fixer ca"""
```

Exercice — Heure 2

```python
mois = ["Jan", "Fev", "Mar", "Avr", "Mai", "Juin"]

ca = [15000, 17000, 16500, 19000, 22000, 25000]
benefice = [3000, 3600, 3400, 4300, 5200, 6500]

produits = ["PC", "Telephone", "Ecran", "Tablette"]
quantites = [70, 150, 90, 110]

experience = [1, 3, 4, 6, 8, 11, 14]
salaire = [4100, 4800, 5300, 6200, 7000, 8500, 10200]

#1
fig,axes = plt.subplots(2,2,figsize = (10,8))
#2
axes[0,0].plot(mois,ca)
#3
axes[0,1].plot(mois,benefice)
#4
axes[1,0].bar(produits,quantites)
#5
axes[1,1].scatter(experience,salaire)
#6
axes[0,0].set_xlabel("mois")
axes[0,0].set_ylabel("CA en DH")
axes[0,0].set_title("Evolution du CA")

axes[0,1].set_xlabel("mois")
axes[0,1].set_ylabel("benefice en DH")
axes[0,1].set_title("Evolution du benefice")

axes[1,0].set_xlabel("produits")
axes[1,0].set_ylabel("quantites")
axes[1,0].set_title("comparaison du quantites par produits")

axes[1,1].set_xlabel("experience")
axes[1,1].set_ylabel("salaire")
axes[1,1].set_title("Relation entre salaire et experience de l'employer")
#7
fig.suptitle("Ma figure exercice 2")
#8
axes[0,0].grid()
axes[0,1].grid()
axes[1,1].grid()
#9
plt.tight_layout(rect = [0,0,1,0.95])
plt.show()
#10
"""Dans les axes : [0,0] et [0,1] on constate que l'evolution du Ca et benefice a augmente de jan a fev,
apres il y a une chutte entre fev et mar , apres il y a un accroissement jusqu'au juin
dans l'axe [1,0] on remarque que le produit : telephone a le plus grande quatite en comparant avec
les autres produits, finalement dans l'axe [1,1] Le nuage de points montre une forte tendance positive :
plus l'expérience  augmente, plus le salaire tend à augmenter."""
```

Exercice de synthèse final — J18

```python
mois = ["Janvier", "Fevrier", "Mars", "Avril", "Mai", "Juin"]
ca = [15000, 17000, 16500, 19000, 22000, 25000]

produits = [
    "Ordinateur portable",
    "Telephone mobile",
    "Ecran externe",
    "Tablette graphique"
]
quantites = [70, 150, 90, 110]

experience = [1, 3, 4, 6, 8, 11, 14]
salaire = [4100, 4800, 5300, 6200, 7000, 8500, 10200]

villes = [
    "Casablanca",
    "Marrakech",
    "Rabat",
    "Tanger"
]
commandes = [220, 160, 180, 140]

#1
fig,axes = plt.subplots(2,2,figsize=(10,8))
#2
axes[0,0].plot(mois,ca)
#3
axes[0,1].bar(produits,quantites)
#4
axes[1,0].scatter(experience,salaire)
#5
axes[1,1].bar(villes,commandes)
#6
axes[0,0].set_xlabel("les mois")
axes[0,0].set_ylabel("CA en DH")
axes[0,0].set_title("Evolution du CA")

axes[0,1].set_xlabel("les produits")
axes[0,1].set_ylabel("les quantites vendues")
axes[0,1].set_title("comparaison du quantites vendues par produits")

axes[1,0].set_xlabel("l'experience")
axes[1,0].set_ylabel("le salaire")
axes[1,0].set_title("Relation entre experience et salaire")

axes[1,1].set_xlabel("les villes")
axes[1,1].set_ylabel("Les commandes")
axes[1,1].set_title("Comparaison du commandes par villes")
#7
axes[0,1].tick_params(axis ="x",rotation = 45,labelsize = 11)
#8
axes[0,0].set_ylim(10000,26000)
axes[0,1].set_ylim(0,160)
axes[1,0].set_xlim(0,15)
axes[1,0].set_ylim(0,11000)
axes[1,1].set_ylim(0,250)
#9
axes[0,0].grid()
axes[1,0].grid()
#10
fig.suptitle("Exercice de synthèse final")
#11
plt.tight_layout(rect=[0,0,1,0.95])
plt.show()
#12
"""Commencer l'axe Y à 130 exagérerait visuellement les différences 
entre les villes et pourrait induire le lecteur en erreur."""
#13
"""Titre generale c'est pour tous la figure , mais le titre du subplot c'est seulement pour
un graphe"""
#14
"""Parfois les noms sont grandes ce qui resulte d'un chevauchement donc on utilise une rotation"""
#15
"""Axe[0,0] : on constate qu'il y a une augmentation du evolution du CA entre jan et fev
apres il y a une legere diminution entre fev et mars , apres mars il y a un accroissement
jusqu'au juin, dans l'axe[0,1] on remarque que les telephones mobiles ont la plus quantite vendue
, passant a l'axe[1,0] la nuage des points montre une relation tres forte positive , plus l'experience 
augmente plus le salaire augmente , finalement l'axe [1,1] on remarque que Casablanca a la plus commandes
en comparant avec les 3 autres villes """
```
